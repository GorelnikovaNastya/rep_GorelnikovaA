2. Для каждого столбца определите оптимальный с точки зрения потребления памяти тип данных - напишите свои выводы в комментариях - 2 балла

**числовые типы**

- столбцы bedrooms/bathrooms/stories/parking  = 6 - диапазон int8

**тип object **

столбцы c YES|NO guestroom/basement/hotwaterheating/airconditioning/prefarea переводим в category

3. Поменяйте типы данных столбцов датафрейма на выбранные вами в прошлом пункте и сравните потребление памяти до и после оптимизации - 1 балл

```python
housing = housing.fillna(0)
```

```python
housing['price'] = housing['price'].astype(np.int32)
housing['area'] = housing['area'].astype(np.int16)
housing['bedrooms'] = housing['bedrooms'].astype(np.int8)
housing['bathrooms'] = housing['bathrooms'].astype(np.int8)
housing['stories'] = housing['stories'].astype(np.int8)
housing['parking'] = housing['parking'].astype(np.int8)
```

```python
housing.info()
```

```python
housing['mainroad'].info(memory_usage="deep")
```

```python
housing['mainroad'] = housing['mainroad'].astype("category")
```

```python
housing['mainroad'].info(memory_usage="deep")
```

```python
housing['mainroad'].unique()
```

```python
housing[:2]
```

```python
housing.info()
```

```python
housing['guestroom'] = housing['guestroom'].astype("category")
housing['basement'] = housing['basement'].astype("category")
housing['hotwaterheating'] = housing['hotwaterheating'].astype("category")
housing['airconditioning'] = housing['airconditioning'].astype("category")
```

```python
housing.info()
```

Выводы:
1) сначала оптимизируем int = числовые типы были уменьшены (все, кроме price) - съекономлено 36% памяти (55,5 КB / 35,3 KB)  
2) оптимизируем object = приведем к категориальным значениям
   - протестировала изменения по столбцу mainroad. До изменения типа memory usage: 27.7 KB, после memory usage: 888.0 bytes
   - перевела столбцы c YES|NO guestroom, basement, hotwaterheating, airconditioning и prefarea в category - общая экономия 69% (55,5 КB / 17,2 КB)

ИТОГО: получена значительная экономия памяти
