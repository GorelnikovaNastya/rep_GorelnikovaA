```python
class Chocolate:
    def __init__(self, price, add=None): #Инициализируем объект, учитываем необязательный аргумент
        if not isinstance(price, float): #Проверяем тип цены
            raise TypeError('Цена не типа float')

        if price <= 0:
            raise ValueError('Цена меньше нуля')

        if add is not None:
            if not isinstance(add, str): #Проверяем тип добавки
                raise TypeError('Добавка не типа string')

        self.price = price
        self.add = add

    def show_my_chocolate(self):
            if self.add is not None and type(self.add) is str: # дублирую проверку по типу добавки на случай замены напрямую
                print(f'Шоколадный батончик, добавка: {self.add}')
            else:
                print('Обычный шоколадный батончик')

    def __str__(self): # Преобразование к строке str
        return f'Шоколадный батончик, {self.price} рублей'
```

```python
chocolate = Chocolate(100.9, 'вишня')
print(chocolate)
chocolate.show_my_chocolate()
```

```python
chocolate = Chocolate(-12, 'куркума')
print(chocolate)
chocolate.show_my_chocolate()
```
