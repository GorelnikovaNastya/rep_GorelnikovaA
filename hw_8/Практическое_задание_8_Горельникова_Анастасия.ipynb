Получить информацию обо всех домах Вестероса (Houses) и сохранить в pd.DataFrame - 1 балл

```python
url2 = "https://anapioficeandfire.com/api/houses"

all_houses = []
page = 1

while True:
    response = requests.get(
        url2,
        params={"page": page, "pageSize": 50}
    )
    response.raise_for_status()

    houses = response.json()
    all_houses.extend(houses)

    # Если следующей страницы нет — заканчиваем
    if "next" not in response.links:
        break

    page += 1

df = pd.DataFrame(all_houses)

print(df.shape)
df.head()
```

Получить информацию обо всех домах Вестероса, у которых есть девиз, и сохранить в pd.DataFrame - 2 балла

```python
url = "https://anapioficeandfire.com/api/houses"
all_houses = []
page = 1

# Получаем все дома по страницам
while True:
    response = requests.get(
        url,
        params={"page": page, "pageSize": 50}
    )
    response.raise_for_status()

    houses = response.json()
    all_houses.extend(houses)

    # Если следующей страницы нет — заканчиваем
    if "next" not in response.links:
        break

    page += 1

# Оставляем только дома с девизом
houses_with_words = [
    house for house in all_houses
    if house.get("words")  # девиз не пустой
]

df = pd.DataFrame(houses_with_words)

print(f"Домов с девизом: {len(df)}")
df[:2]
```
